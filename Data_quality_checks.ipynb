```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("Data Quality")
    .master("local[*]")
    .getOrCreate()
)
```

```python
sales_df = spark.read.parquet(
    "output/gold/sales_fact"
)

```

```python

print("Null Customer IDs")
print(
    sales_df.filter(
        sales_df.customer_id.isNull()
    ).count()
)
```

```python
print("Negative Revenue")
print(
    sales_df.filter(
        sales_df.payment_value < 0
    ).count()
)
```

```python
print("Duplicate Records")

print(
    sales_df.count()
    -
    sales_df.dropDuplicates().count()
)
```
